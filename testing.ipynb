```python
import tkinter as tk
from tkinter import filedialog
from PIL import Image, ImageTk
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import img_to_array

# -----------------------------
# LOAD MODEL
# -----------------------------
MODEL_PATH = "healthy_vs_rotten.h5"
model = load_model(MODEL_PATH, compile=False)

# -----------------------------
# CLASS NAMES (28 CLASSES)
# ⚠️ ORDER MUST MATCH MODEL TRAINING
# -----------------------------
class_names = [
    "Apple_healthy",
    "Apple_rotten",
    "Banana_healthy",
    "Banana_rotten",
    "Bellpepper_healthy",
    "Bellpepper_rotten",
    "Carrot_healthy",
    "Carrot_rotten",
    "Cucumber_healthy",
    "Cucumber_rotten",
    "Grape_healthy",
    "Grape_rotten",
    "Guava_healthy",
    "Guava_rotten",
    "Mango_healthy",
    "Mango_rotten",
    "Orange_healthy",
    "Orange_rotten",
    "Potato_healthy",
    "Potato_rotten",
    "Strawberry_healthy",
    "Strawberry_rotten",
    "Tomato_healthy",
    "Tomato_rotten",
    "Watermelon_healthy",
    "Watermelon_rotten",
    "Papaya_healthy",
    "Papaya_rotten"
]

IMG_SIZE = 100

def upload_image():
    file_path = filedialog.askopenfilename(
        filetypes=[("Image files", "*.jpg *.jpeg *.png")]
    )
    if not file_path:
        return

    img = Image.open(file_path).resize((IMG_SIZE, IMG_SIZE))
    img_display = ImageTk.PhotoImage(img)

    image_label.config(image=img_display)
    image_label.image = img_display

    predict_image(file_path)

def predict_image(image_path):
    img = Image.open(image_path).resize((IMG_SIZE, IMG_SIZE))
    img_array = img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) / 255.0

    prediction = model.predict(img_array)
    class_index = np.argmax(prediction)
    confidence = np.max(prediction) * 100

    result_label.config(
        text=f"Prediction: {class_names[class_index]}\nConfidence: {confidence:.2f}%"
    )

# -----------------------------
# GUI WINDOW
# -----------------------------
root = tk.Tk()
root.title("Healthy vs Rotten Fruit & Vegetable Classifier")
root.geometry("420x520")
root.resizable(False, False)

title = tk.Label(
    root,
    text="Fruit & Vegetable Quality Detection",
    font=("Arial", 14, "bold")
)
title.pack(pady=10)

image_label = tk.Label(root)
image_label.pack(pady=10)

upload_btn = tk.Button(
    root,
    text="Upload Image",
    command=upload_image,
    font=("Arial", 12),
    bg="green",
    fg="white",
    width=18
)
upload_btn.pack(pady=10)

result_label = tk.Label(
    root,
    text="Prediction will appear here",
    font=("Arial", 12)
)
result_label.pack(pady=20)

root.mainloop()
```
